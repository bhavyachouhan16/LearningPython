```python
# Question 1
username=input("Enter Username : ")
username=username.strip(" ")
if len(username)<5 or " " in username:
    print("Invalid Username")
else:
    print("Valid Username")
```

```python
# Question 2
email=input("Enter Email : ")
email=email.split("@")
print(email)
print("Domain is : ",email[1])
```

```python
# Question 3
password=input("Enter Password : ")
is_upper=False
is_digit=False
for i in password:
    if i.isdigit():
        is_digit=True
    elif i.isupper():
        is_upper=True
if len(password)>8 and is_digit and is_upper:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
# Question 4
s=input("Enter String : ")
s=s.lower()
vowel_count=0
consonant_count=0
for i in s:
    if i in 'aeiou':
        vowel_count+=1
    elif i not in 'aeiou' and i!=" ":
        consonant_count+=1
print(vowel_count)
print(consonant_count)
    
```
