```python
# TUPLE :- It has collection of comma separated data in round bracket()
# It is ordered,immutable,heterogenius(multiple datatype),indexed

t=tuple()       # empty tuple
print(type(t))

a=(10,)
print(type(a))

b=[10,20,30]
print(b)

c=tuple(b)
print(c)
```

```python
t=(10,"aliya",10.5,100,200)
print(t[0])    # indexing
print(t[2])
t[2]="DA"      # error because tuple is immutable
```

```python
# slicing 
t=(10,"aliya",10.5,[100,200],55,65)
print(t[1::])
```

```python
# looping through tuple
# direct looping 
# indirect looping (using range())

t=(10,"aliya",10.5,100,200)    # direct looping
for i in t:
    print(i,end=" ")
print()

print("----------")

for i in range(len(t)):
    print(t[i],end=" ")       # indirect looping
```

```python
# tuple concatation
t1=(10,20,20)
t2=(100,200,300)
t=t1+t2
print(t)
```

```python
# tuple packing :- It means pack multiple data in tuple

t=10,10.5,"gagan"          # tuple packing
print(type(t))
print(t)
```

```python
t=10,10.5,"gagan"         # tuple unpacking
name,point,rollnumber=t
print(name)
print(point)
print(rollnumber)
```

```python
# len() :- To find length of tuple

t=(10,20,30,40)
count=0

for i in t:
    count+=1
print(count)

print(len(t))
```

```python
# max() :- To find maximum value in tuple

t=(10,20,30,40)
largest=t[0]
for i in t:
    if i>largest:
        largest=i
print(largest)
   
print(max(t))

```

```python
# min :- To find minimum value in tuple

t=(10,20,30,40)
smallest=t[0]
for i in t:
    if i<smallest:
        smallest=i
print(smallest)

print(min(t))
```

```python
# sum :- give total value of tuple

t=(10,20,30,40)
s=0
for i in t:
    s+=i
print(s)

print(sum(t))
```

```python
# sorted :- Arrange tuple (By default ascending order)

t=(20,30,10,40)
print(sorted(t))                # Arranged tuple in ascending order)
print(sorted(t,reverse=True))   # Reverse Tuple
```

```python
# Typecasting
# tuple to list

t=(20,30,10,40)
a=list(t)
print(a)
print(type(a))

# list to tuple
t=[20,30,10,40]
a=tuple(t)
print(a)
print(type(a))
```

```python
# TUPLE METHODS
# count() :- To find occurance of particular element

t=(10,10,20,20,20,30,40,50)
print(t.count(20))
```

```python
# index(value,(index)) :- To find index value of any particular element
                       # 2nd argument is optional
t=(10,10,20,20,20,30,40,50)
print(t.index(20))
print(t.index(20,3))
```

```python
# enumerate() :- It gives index-value pair
# enumerate(tuple,start)

t=(10,10,20,20,20,30,40,50)
for i in enumerate(t):
    print(i)

print("-------------")

t=(10,10,20,20,20,30,40,50)
for i in enumerate(t,start=5):      # 2nd argument is optional 
    print(i)
```

```python
# zip() :- It is used to combine multiple tuple
names=("Ruchi","Bhavya","Alia","Laxmi")
courses=("DA","DS","ML","BA")
fees=(25000,40000,45000,30000)
for i in zip(names,courses,fees):
    print(i)
```
