```python
# index :- position of a character in a given string
# first character of any string is at 0 index.
# last character is at -1 index.
# indexing :- it is a technique for accessing individual character from a given string.

s="welcome to bhopal"
print(s[0])
print(s[-1])
print(s[7])
print(s[-4])
```

```python
# slicing :- it is a techinque for extracting substring from a given string.
# strobj[start:stop:step]
# start - by default - 0 index
# stop - by default end of the string or n-1
# step - by default - +1 increment 
s="welcome to bhopal"
print(s[::])
print(s[:7:])
print(s[::-1])      # start -> -1  # stop -> end of the string ---> reverse string
print(s[-1:-7:-1])
```

```python
s="welcome to city of lakes"
print(s[1:15:2])
print(s[-2:-16:-2])
print(s[:11:])
print(s[:7:2])
print(s[-1:-14:-2])
print(s[1::3])
```

```python
# wap to print number of vowels , number of consonants and number of spaces in a given string?
s="welcome to city of lakes"
c_vowel=0
c_consonant=0
c_space=0
for i in s:
    if i in ("aeiouAEIOU"):
        c_vowel=c_vowel+1
    elif i==" " :
        c_space=c_space+1
    else:
        c_consonant=c_consonant+1
       

print("count of vowel:",c_vowel)
print("count of consonant:",c_consonant)
print("count of space:",c_space)
```

```python
# wap to print reverse of any given string?
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
print(rev)      # without slicing
print(s[::-1])  # with slicing
```

```python
# wap to check whether a given string is palindrome or not?
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
if s==rev:
    print(s,"is palindrome string")
else:
    print(s,"not a palindrome string")
```

```python
# count() :- return occurrence of any character in a given string.
# strobj.count(char/substring,[start],[stop])
# start - by default - 0
# stop - by default - end of the string
# stop - n -----> (n-1)
s="welcome to bhopal"
print(s.count("o"))
print(s.count('o',5))
print(s.count(' '))
print(s.count('bhopal'))
print(s.count('z'))
print(s.count('o',5,9))
```

```python
# index() :- return position of character/substring from a given string.
# strobj.index(char/substring,[start],[stop])
# always return first occurrence
# index through error if substring is not found
s="welcome to bhopal"
print(s.index('o'))
print(s.index('o',))
print(s.index('bhopal'))
print(s.index('o',5,9))
```

```python
# find() :- return position of character/substring from a given string.
# strobj.find(char/substring,[start],[stop])
# always return first occrrence.
# find returns -1 if substring is not found 
s="welcome to bhopal"
print(s.find("o"))
print(s.find('o',5))
print(s.find('bhopal'))
print(s.find('o',5,9))
```
