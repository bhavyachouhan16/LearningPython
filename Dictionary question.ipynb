```python
# Question 11 Count frequency of each character in a string.
s="decodedataacademy"
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
# Question 12 Count frequency of each word in a sentence
s="this is the decode the data the academy"
s=s.split()
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```
