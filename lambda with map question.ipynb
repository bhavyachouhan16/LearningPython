```python
# QUESTION 1: DOUBLE EVERY NUMBER
numbers = [1, 2, 3, 4, 5]
double=map(lambda x:x+x,numbers)
print(list(double))
```

```python
# QUESTION 2: ADD 10 TO EVERY NUMBER
numbers = [5, 10, 15, 20]
add=map(lambda x:x+10,numbers)
print(list(add))
```

```python
# QUESTION 3: SUBTRACT 5 FROM EVERY NUMBER
numbers = [10, 20, 30, 40]
sub=map(lambda x:x-5,numbers)
print(list(sub))
```

```python
# QUESTION 4: MULTIPLY EVERY NUMBER BY 3
numbers = [2, 4, 6, 8]
multi=map(lambda x:x*3,numbers)
print(list(multi))
```

```python
# QUESTION 5: FIND THE SQUARE OF EVERY NUMBER
numbers = [1, 2, 3, 4, 5]
square=map(lambda x:x**2,numbers)
print(list(square))
```

```python
# QUESTION 6: FIND THE CUBE OF EVERY NUMBER
numbers = [1, 2, 3, 4]
cube=map(lambda x:x**3,numbers)
print(list(cube))
```

```python
# QUESTION 7: CONVERT NUMBERS TO STRINGS
numbers = [10, 20, 30, 40]
string=map(lambda x:str(x),numbers)
print(list(string))
```

```python
# QUESTION 8: FIND THE REMAINDER AFTER DIVIDING BY 5
numbers = [10, 12, 17, 23, 30]
remainder=map(lambda x:x%5,numbers)
print(list(remainder))
```

```python
# QUESTION 9: INTEGER DIVISION BY 2
numbers = [10, 15, 20, 25]
floor=map(lambda x:x//2,numbers)
print(list(floor))
```

```python
# QUESTION 10: CONVERT CELSIUS TO FAHRENHEIT
temperatures = [0, 10, 20, 30, 40]
fahr=map(lambda x:x*9/5+32,temperatures)
print(list(fahr))
```

```python
# QUESTION 11: CONVERT METERS TO CENTIMETERS
meters = [1, 2, 5, 10]
centimeter=map(lambda x:x*100,meters)
print(list(centimeter))
```

```python
# QUESTION 12: ADD GST OF 18%
prices = [100, 200, 500, 1000]
final_price=map(lambda x:x+(x*18/100),prices)
print(list(final_price))
```

```python
# QUESTION 13: CONVERT NAMES TO UPPERCASE
names = ["gagan", "rahul", "amit", "priya"]
upper=map(lambda x:x.upper(),names)
print(list(upper))
```

```python
# QUESTION 14: CONVERT NAMES TO LOWERCASE
names = ["GAGAN", "RAHUL", "Amit", "PRIYA"]
lower=map(lambda x:x.lower(),names)
print(list(lower))
```

```python
# QUESTION 15: FIND LENGTH OF EVERY WORD
words = ["Python", "Django", "SQL", "React"]
length=map(lambda x:len(x),words)
print(list(length))
```

```python
# QUESTION 16: ADD PREFIX TO EVERY NAME
names = ["Rahul", "Amit", "Gagan"]
prefix=map(lambda x:"Mr. "+x,names)
print(list(prefix))
```

```python
# QUESTION 17: ADD PREFIX TO EVERY CITY
cities = ["Bhopal", "Indore", "Jaipur", "Delhi"]
prefix_city=map(lambda x:"City :"+x,cities)
print(list(prefix_city))
```

```python
# QUESTION 18: CALCULATE SIMPLE INTEREST FOR EACH RECORD
principal = [1000, 2000, 5000, 10000]
si=map(lambda x:x*5*2/100,principal)
print(list(si))
```

```python
# QUESTION 19: CALCULATE AREA OF SQUARES
sides = [2, 4, 5, 10]
area_square=map(lambda x:x*x,sides)
print(list(area_square))
```

```python
# QUESTION 20: CONVERT MINUTES TO SECONDS
minutes = [1, 5, 10, 30]
seconds=map(lambda x:x*60,minutes)
print(list(seconds))
```
