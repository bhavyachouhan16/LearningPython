```python
# wap to print 1st n natural number?
n=int(input("enter any number: "))
i=1
while i<n+1:
    print(i)
    i=i+1
```

```python
# wap to print sum of 1st n natural number?
n=int(input("enter any number: "))
i=1
sum=0
while i<n+1:
    sum=sum+i
    i=i+1
print(sum)
```

```python
# wap to print sum and count of 1st n even number and sum and count of 1st n odd number between range start and end?
# take 2 user input start and end.
start=int(input("enter any number: "))
end=int(input("enter any number: "))
i=start
sum_even=0
sum_odd=0
c_even=0
c_odd=0
while i<end+1:
    if i%2==0:
        sum_even=sum_even+i
        c_even=c_even+1
    else:
        sum_odd=sum_odd+i
        c_odd=c_odd+1
    i=i+1
print(sum_even,c_even)
print(sum_odd,c_odd)
```

```python
# wap to print reverse of any given number ?
n=int(input("enter any number: "))
rev=0
while n!=0:
    dig=n%10          # last digit
    rev=rev*10+dig    # output
    n=n//10           # loop
print(rev)
```

```python
# wap to print reverse number between start and end ?
# take start and end as a user input at a runtime.
start=int(input("enter any number: "))
end=int(input("enter any number: "))
for i in range(start,end+1):
    rev=0
    while i!=0:
        dig=i%10
        rev=rev*10+dig
        i=i//10
    print(rev)
```

```python
# Question 1. Find the sum of digits of a number.
n=int(input("enter any number: "))
sum=0
while n!=0:
    dig=n%10
    sum=sum+dig
    n=n//10
print(sum)
```

```python
# Question 2. Find the product of digits.
n=int(input("enter any number: "))
pro=1
while n!=0:
    dig=n%10
    pro=pro*dig
    n=n//10
print(pro)
```

```python
# Question 3. Find the largest digit.
n=int(input("enter any number: "))
max=0
while n!=0:
    dig=n%10
    if dig>max:
        max=dig
    n=n//10
print(max)
```

```python
# Question 4. Find the smallest digit.
n=int(input("enter any number: "))
min=9
while n!=0:
    dig=n%10
    if dig<min:
        min=dig
    n=n//10
print(min)
```
