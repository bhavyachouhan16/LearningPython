```python
Q1=np.quantile(df_app["AMT_ANNUITY"].dropna(),0.25)
Q1
```

```python
Q2=np.quantile(df_app["AMT_ANNUITY"].dropna(),0.50)
Q2
```

```python
Q3=np.quantile(df_app["AMT_ANNUITY"].dropna(),0.75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower_fence=Q1-(1.5*IQR)
lower_fence
```

```python
upper_fence=Q3+(1.5*IQR)
upper_fence
```

```python
l=[]
for i in df_app["AMT_ANNUITY"]:
    if i>upper_fence or i<lower_fence:
        l.append(i)
print(len(l))
```

```python
plt.figure(figsize=(12,8))
name_type_suite_count=df_app["NAME_TYPE_SUITE"].value_counts()
sns.barplot(x=name_type_suite_count,y=name_type_suite_count.index,palette="viridis")
plt.title("NAME_TYPE_SUITE BAR PLOT")
plt.xlabel("FREQUENCY")
plt.ylabel("NAME_TYPE_SUITE")

for index,value in enumerate(name_type_suite_count):
    plt.text(value,index,f'{value}',ha="left",va="center",color="black",fontsize=10,weight="normal")
plt.show()
```
