```python
# Numeric Datatype
## Integer :- identify to + identify,unindexed,immutable
## Float:- infinite to + infinite,continous manners

### type()- it is used to check the datatype of variable

a=10
print(type(a))

b=10.5
print(type(b))
```

```python
# String Datatype
## definition :- character inside double quote or single quote
## Single line :- single quote ('') or double quote ("")
## Multiple line :- (''' ''') or (""" """)
# it is indexed,immutable,heterogenious

s1="Gagan123"   #single line string
s2="""
    My
    Name
    Is 
    Gagan
    """
print(s1)
print(s2)
```

```python
# Sequence 
# list - collection of heterogenious datatype with coma separate values
## indexed,mutable,sequence 

a=[10,20,10.5,"Gagan"]
print(a)
print(a[3])   #positive indexing
print(a[-1])  #negative indexing
a[2]=50
print(a[3][2])
```

```python
# tuple - it is collection of element inside round bracket() with coma separate
## it is indexed,immutable,sequence 
## it is execute faster than list
## we cannot remove elements from tuple after creation

t=(10,20,10.5,"Gagan")
print(t)
print(type(t))
```

```python
# Mappping
## Dictionary
## it is collection of data in key and value pair  key:value 
## unindexed,mutable,key should be unique and data should be duplicate
a={"name":["yash","ruchi","disha"],"age":22}
print(a)
print(a.get("name"))
print(a.get("age"))
```

```python
# Set
## set,frozen set
## it is collection of unique data,heterogenious,unindexed 
## it is created using{}
## union, intersection
s={10,20,20,30,"Gagan"}
print(s)
print(type(s))
print(s)

#frozenset is immutable
fs=frozenset(s)
print(type(fs))
```

```python
# Complex number
c=3+4J
print(type(c))
```

```python
# find factorial of number
# Reverse loop
n=5
mul=1
for i in range(n,0,-1):
    mul=mul*i
print(mul)
```

```python
# Forward loop
n=5
mul=1
for i in range(1,n+1):
    mul=mul*i
print(mul)
```
