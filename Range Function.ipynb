```python
# Range Function :-  it gives us sequence of number
## range(start,stop,step)
# we have to pass atleast one argument which is stop

for i in range(5):
    print(i)
   
```

```python
for i in range(1,10,2):
    print(i)
```

```python
# Reverse

for i in range(10,0,-2):
    print(i)
```

```python
# print square of 1 to 10 using range()
## use exponent operator(**)
for i in range(1,11,1):    #(1,11)
  print(i**2)              # i**2 or i*i
```
