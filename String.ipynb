```python
# String :- It is a sequence of character
          # character should be a-z,A-Z,Numeric value,special symbol
# it is written in Single quote or double quote
# indexed,immutable

s="Gagan"   # Single line String
print(s[0])
print(s[1])
print(s[2])
print(s[3])
print(s[4])
```

```python
# String function
## function_name(string)
## len(),max(),min(),sorted()

s="DecodeDataAcademy"
sum=0
for i in s:
    sum=sum+1
print(sum)   # length without function
print(len(s)) # lenghth with function
```

```python
## max(),min()
## ord() - to find ascii value of character
## chr() - covert Ascii value to character

s="Gagan"
for i in s:
    print(i,ord(i))
print("Maximum is : ",max(s))
print("Minimum is : ",min(s))

print(chr(97))
```

```python
# Find Maximum of string without using max() function
s="Gagan"
largest=s[0]
for i in range(len(s)):
    if s[i]>largest:
        largest=s[i]

print(largest)
print(max(s))
```

```python
# Find Minimum of string without using min() function
s="Gagan"
smallest=s[0]
for i in range(len(s)):
    if s[i]<smallest:
        smallest=s[i]
print(smallest)
print(min(s))
```

```python
# sorted() - to arrange string in order in list form
s="Gagan"
print(sorted(s))
print(sorted(s,reverse=True))      # for decending order
```

```python
# String Methods
## string.function_name()
s="DecOde daTa acaDemy"
print(s.upper())
print(s.lower())
print(s.capitalize())
print(s.title())
print(s.swapcase())
```

```python
# strip(),lstrip(),rstrip()
s="@Gagan@"
print(s)
print(s.strip("@"))
print(s.lstrip("@"))
print(s.rstrip("@"))
```

```python
# replace("old","new") method -
s="Decode ata Academy"
print(s)
a=s.replace("ata","Data")
print(a)

s1="decode"
s2=s1.replace("o","a")
print(s2)

l=["a","b","c"]
l[0]="x"
print(l)
```

```python
# find() method - if find substring then output will index position
               # if find substring then output will index position
s="Decode Data Acedemy"
print(s.find("Data"))
print(s.find("data"))
```

```python
# index() method - if find substring then output will first occurance index position
                # if find substring then error
s="Decode Data Academy"
print(s.index("Data"))
# print(s.index("data"))    # error
s="Decode"
print(s.index("e",2))
```

```python
# count() method - if find count of occurance of substring
s="Decode Data Academy"
print(s.count("Data"))
print(s.count("data"))
```

```python
# startswith() method -
# endswith() method -
s="Decode Data Academy"
print(s.startswith("D")) # True
print(s.startswith("d")) # False
print(s.endswith("y"))   # True
```

```python
# split() - to do part string - It gives list after split
s="Decode Data Academy"
a=s.split(" ")
print(a)
s="Decode#Data#Academy"
a=s.split("#")
print(a)
```

```python
# join() - to join iterables into string
## format - " ".join(iterables)
l=["Decode","Data","Academy"]
s=" ".join(l)
print(s)
```

```python
# {extra}
l1=["Decode","Data","Academy"]
l2=["Ruchi","Yash"]
l=l1.extend(l2)
s=" ".join(l1)
print(s)
```
