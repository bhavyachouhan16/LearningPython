```python
class Account:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance

account1=Account("Gagan",50000)
print(account1.name)
print(account1.balance)
account1.balance-=5000
print(account1.balance)
```

```python
##                     PILLERS OF OOPs

# 1. Encapsulation :-
# Wraping data and method is single unit(class) and restrict direct access of object's
# data outside of class.
# Access Specifier :- Provide security to object data
   # 1. Public variable :- self.name=name
   # 2. Protected variable :- self._name=name
   # 3. Private variable :- self.__name=name
## getter method and setter method :- getter ansd setter method used to set or get private varibale
    # or used to perform operations on private variable

class Account:
    def __init__(self,name,balance):
        self.name=name         # public variable
        self.__balance=balance # private variable

    def getbalance(self):
        print("Balance is ",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount
    
a1=Account("Gagan",5000)
print(a1.name)
a1.getbalance()
a1.setbalance(20000)
a1.getbalance()
```

```python
# Method 1 of Getter and Setter

class Account:
    def __init__(self,name):
        self.__name=name

    def get_name(self):          # Traditional Getter Method
        print(self.__name)

    def set_name(self,newname):  # Traditional Setter Method
        self.__name=newname

a=Account("Gagan")
a.get_name()
a.set_name("Ruchi")
a.get_name()
```

```python
# Method 2 of Getter and Setter
# We use @property decorator

class Account:
    def __init__(self,name):
        self.__name=name

    @property
    def name(self):            # Getter Method    
        print(self.__name)

    @name.setter
    def name(self,newname):   # Setter Method
        self.__name=newname

a=Account("Gagan")
print(a.name)
a.name="Ruchi"
print(a.name)
```
