```python
# broadcasting in numpy is powerful technique thats allow mathematical operation between
# two array of different shape and dimension 

# Broadcasting Rules:-

# when operating two array numpy evaluate their shape elementwise starting from rightmost to the leftmost.If two dimension
# array are compatible.

# 1. they are same.

# 2. if dimension are different then one of them is exactly one if two array of different dimension 
# the shape of smalller array is left padded with ones.
# ex:- arr1 -> shape(3,) -> 1D       arr2 -> shape(2,3) -> 2D

# 3. Compatibility error rule :- If any dimension array size disagree and neither size equal to 1 broadcasting fails and 
# raise a value error
```

```python
# Join two array
# 1. Concatenate :- always return same dimension array with different 
# shape(increase shape)
# dimension and shape both are same
arr1=np.array([10,20,30,40])
arr2=np.array([100,200,300,400])
print(arr1)
print(arr2)
```

```python
arr_new=np.concatenate((arr1,arr2))
print(arr_new)
```
