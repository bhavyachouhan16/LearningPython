```python
# Armstrong Number :- 
# A number that equals to sum of its own digit , each raised to the power of the total numbers of digits.
```

```python
# wap to check whether a given a number is an armstrong number or not?
n=int(input("enter any number: "))
a=len(str(n))
sumdigit=0
old=n
while n!=0:
    dig=n%10
    sumdigit=sumdigit+(dig**a)
    n=n//10
if sumdigit==old:
    print(old,"Armstrong Number")
else:
    print(old,"Not an Armstrong Number")
```

```python
# wap that print armstrong number between start and end .
# take start and end as a user input as a runtime. 1 to 2000
start=int(input("enter any number: "))
stop=int(input("enter any number: "))
for i in range(start,stop+1):
    sm=0
    a=len(str(i))
    old=i
    while i!=0:
        dig=i%10
        sm=sm+(dig**a)
        i=i//10
    if sm==old:
        print(old,end=" ") 
```
