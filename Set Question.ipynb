```python
# Question 1
# Create a set of five integers
s={10,20,20,30,40,40,50,50}
print(s)
print(type(s))
```

```python
# Question 2
# Create an empty set and print its type
s=set()
print(type(s))
```

```python
# Question 3
# Add a new element to a set using add()
s={1,2,3,4,5}
s.add(6)
print(s)
```

```python
# Question 4
# Add multiple elements using update()
s={1,2,3,4,5}
s.update({6,7})
print(s)
```

```python
# Question 5
# Remove an element using remove()
s={1, 2, 3, 4, 5, 6, 7}
s.remove(6)
print(s)
```

```python
# Question 6
# Remove an element using discard()
s={1,2,3,4,5,6}
s.discard(7)
print(s)
```

```python
# Question 7
# Remove and print a random element using pop()
s={1,2,3,4,5,6}
s.pop()
print(s)
```

```python
# Question 8
# Find the length of a set
s={10,20,20,30,40,40,50,50}
print(len(s))
```

```python
# Question 9
# Check whether a value exists in a set using the ‘in’ operator
s={10,20,30,40,50}
print(20 in s)
```

```python
# Question 10
# Convert a list containing duplicate values into a set
l=[10,20,20,30,30,40,40,50,50]
s=set(l)
print(s)
print(type(s))
```

```python
# Question 11
# Find the union of two sets
s1={1,2,3,4,5}
s2={6,7,8,9,10}
s=s1.union(s2)
print(s)
```

```python
# Question 12
# Find the intersection of two sets
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.intersection(s2)
print(s)
```

```python
# Question 13
# Find the difference between two sets
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.difference(s2)
print(s)
```

```python
# Question 14
# Find the symmetric difference between two sets
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.symmetric_difference(s2)
print(s)
```

```python
# Question 15
# Check whether one set is a subset of another.
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.issubset(s2)
print(s)
```

```python
# Question 16
# Check whether one set is a superset of another
s1={1,2,3}
s2={1,2,3,4,5,6,7,8}
s=s2.issuperset(s1)
print(s)
```

```python
# Question 17
# Check whether two sets are disjoint
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.isdisjoint(s2)
print(s)
```

```python
# Question 18
# Count the number of unique words in a sentence using a set.
s="progamming"
a=set(s)
print(len(a))
```

```python
# Question 19
# Find common elements between two lists using sets
l1=[1,2,3,4,5]
s1=set(l1)
l2=[4,5,6,7,8]
s2=set(l2)
a=s1.intersection(s2)
print(a)
```
