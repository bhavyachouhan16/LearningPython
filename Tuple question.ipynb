```python
# part 2
# Question 11
# print even number
t=(2,3,5,6,8,10,12,14)
for i in t:
    if i%2==0:
        print(i,end=" ")
```

```python
# Question 12
# count even and odd number
t=(2,3,5,6,7,8,9,11,16,15,14,18,17,19)
even=0
odd=0
for i in t:
    if i%2==0:
        even+=1
    else:
        odd+=1
print("even number",even)
print("odd number",odd)
```
