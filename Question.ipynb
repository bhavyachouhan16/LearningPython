```python
# Question 1
# greatest among 3 numbers
a=int(input("Enter number a : "))
b=int(input("Enter number b : "))
c=int(input("Enter number c : "))
if a>b and a>c:
    print(a,"is greatest")
elif b>a and b>c:
    print(b,"is greatest")
else:
    print(c,"is greatest")
```

```python
# Question 3
# leap year
## year should be divisible by 400 or year divisible by 4 and not divisible by 100
year=int(input("Enter year to check :"))
if year % 4 == 0:
    if year % 100 ==0:
        if year % 400 ==0:
            print("This is leap year")
        else:
            print("Not leap year")
    else:
        print("leap year")
else:
    print("Year is not leap year")
```

```python
## leap year 
## another method to check leap year or not 

year=int(input("Enter year to check :"))
if year % 400==0 and year % 100!=0:
    print("Year is leap year")
else:
    print("Not leap year")
```

```python
# Question 8
# triangle type
a=int(input("enter one side a : "))
b=int(input("enter second side b : "))
c=int(input("enter third side c :"))
if a==b and a==c and b==c:
    print("Equilateral triangle")
elif (a==b and b!=c) or (b==c and b!=a) or (a==c and a!=b):
    print("Isosceles triangle")

else:
    print("Scalene triangle")

            
```
