```python
# Q3. arange() + reshape() — Logic Create an array containing numbers from 10 to 69 using arange().
# Reshape it into a 3D array of shape (4, 5, 3).
#Then:
#1. Find the sum of the elements along the last axis.
#2. Find the maximum value from each 5 × 3 array.
#3. Extract the third 5 × 3 array.
#4. Replace all values divisible by 7 with -1.

arr=np.arange(10,70).reshape(4,5,3)
print(arr)
print()
print(np.sum(arr,axis=2))
print()
print(np.max(arr,axis=0))
print()
print(arr[2])

for i in np.nditer(arr):
    if i%7==0:
        i=-1
print(arr)
```

```python
# Q5. Random Array + Quartiles Generate a random NumPy array containing 50 integers between 10 and 100.
#Find:
#• Q1
#• Median
#• Q3
#• IQR
#• Lower Fence
#• Upper Fence

arr=np.random.randint(10,100,50)
print(arr)
print(np.median(arr))
```

```python
Q1=np.quantile(arr,0.25)
Q1
```

```python
Q3=np.quantile(arr,0.75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower_fence=Q1-(1.5*IQR)
lower_fence
```

```python
upper_fence=Q3+(1.5*IQR)
upper_fence
```
