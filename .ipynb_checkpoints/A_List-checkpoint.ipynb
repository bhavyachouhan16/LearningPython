```python
# list - collection of an element denoted by [] barcket with coma separated value
# indexed,ordered,mutable and heterogenous in nature.
l=[]     # empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l)
print(type(l))
print(l[0])    # indexed
l[2]="decode"  # mutable
print(l)
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l[-2][-1])
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l[3])
print(l[::])
print(l[::-1])   # reverse list
print(l[1:6:2])
```
