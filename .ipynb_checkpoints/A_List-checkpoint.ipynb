```python
# list - collection of an element denoted by [] barcket with coma separated value
# indexed,ordered,mutable and heterogenous in nature.
l=[]     # empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l)
print(type(l))
print(l[0])    # indexed
l[2]="decode"  # mutable
print(l)
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l[-2][-1])
```

```python
l=[10,20,30,60,10.5,"hello",10]
print(l[3])
print(l[::])
print(l[::-1])   # reverse list
print(l[1:6:2])
```

```python
# iteration
# direct iteration
l=[10,200,"hello",20,10.5]
print(l)
for i in l:
    print(i)
```

```python
# indirect iteration
l=[10,200,"hello",20,10.5]
for i in range(len(l)):
    print(i,l[i])
```

```python
# Functions
# len(listobj) :- return no of elements in a given list.
l=[10,200,"hello",20,10.5]
print(len(l))
count=0
for i in l:
    count=count+1
print(count)
```

```python
# max(listobj)
l=[10,200,100,10.5,7]
print(max(l))
print(min(l))
```

```python
# max()
l=[20,8,29,9]
max_value=l[0]
for i in l:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
# min()
l=[20,9,87,13]
min_value=l[0]
for i in l:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
# sum() 
l=[20,9,17,87,19]
print(sum(l))
sum=0
for i in l:
    sum=sum+i
print(sum)
```

```python
# mean() :- sum of observation divided by number of observation
l=[20,9,17,87,19]
sum=0
count=0
for i in l:
    sum=sum+i
    count=count+1
mean=sum/count
print(mean)
```

```python
# Methods
# add element to an exiting list
# 1. append
# listobj.append(element)
# by default -> -1
l=[10,20,100,10.5,'hello',78]
l.append(1000)
print(l)
```

```python
l=[10,20,100,10.5,'hello',78]
l.append("bhopal")
print(l)
```

```python
l=[10,20,100,10.5,'hello',78]
l.append([1000,2000,3000,"bhopal"])
print(l)
print(l[-1][1])
print(l[4][1])
print(l[-1][-1][2])
```

```python
l=[10,20,30,[100,200,["abc","def"],300,400],10.5,1000]
print(l)
print(l[3])
print(l[3][2])
print(l[3][2][-1])
```

```python
l=[10,20,30,[100,200,["abc","def"],300,400],10.5,1000]
# [10,20,30,[100,200,["abc","def","decode"],300,400],10.5,1000]
l[3][2].append("decode")
print(l)
```

```python
l1=[10,20,30,[100,200,["abc","def"],300,400],10.5,1000]
l2=["bpl","gwl","jbp"]
#[10,20,30,[100,200,["abc","def","bpl,"gwl","jbp"],300,400],10.5,1000]
for i in l2:
    l1[3][2].append(i)
print(l1)
```
