```python
# list - collection of an element denoted by [] barcket with coma separated value
# indexed,ordered,mutable and heterogenous in nature.
l=[]     # empty list
print(l)
print(type(l))
```

```python
# iteration
# direct iteration
l=[10,200,"hello",20,10.5]
print(l)
for i in l:
    print(i)
```

```python
# indirect iteration
l=[10,200,"hello",20,10.5]
for i in range(len(l)):
    print(i,l[i])
```

```python
# Functions
# len(listobj) :- return no of elements in a given list.
l=[10,200,"hello",20,10.5]
print(len(l))
count=0
for i in l:
    count=count+1
print(count)
```

```python
# max(listobj)
l=[10,200,100,10.5,7]
print(max(l))
print(min(l))
```

```python
# max()
l=[20,8,29,9]
max_value=l[0]
for i in l:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
# min()
l=[20,9,87,13]
min_value=l[0]
for i in l:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
# sum() 
l=[20,9,17,87,19]
print(sum(l))
sum=0
for i in l:
    sum=sum+i
print(sum)
```

```python
# mean() :- sum of observation divided by number of observation
l=[20,9,17,87,19]
sum=0
count=0
for i in l:
    sum=sum+i
    count=count+1
mean=sum/count
print(mean)
```

```python
l=["madam","bhopal","Naman","decode","Ai","nitin"]
# ["madam","nitin"]
l1=[]
for i in l:
    rev=''
    for n in i:
        rev=n+rev
    if rev==i:
        l1.append(i)
print(l1)
```

```python
# index :- return position of an element 
# listobj.index(element,[start],[stop])
# start - by default -> 0
# end - by default -> end of the element - n --> n-1
# always return first occurance
l=[10,20,"hello",100,10,20,200,10,100,10.5]
print(l.index(10))
print(l.index(10,1))
print(l.index(10,5,7))
```

```python
# count() :- return occurance of given element in a list
# listobj.count(element)
l=[10,20,"hello",100,10,20,200,10,100,10.5]
print(l.count(10))
print(l.count(1000))
```

```python
# reverse
# listobj.reverse()
# change existing list not created new list
l=[10,20,"hello",200,9]
l.reverse()
print(l)
```

```python
# slicing reverse
# created a new list
l=[10,20,"hello",200,9]
print(l[::-1])
```

```python
l=[10,20,100,40]
# [40,100,20,10]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        l[i],l[j]=l[j],l[i]
print(l)
```

```python
# Ascending order
l=[10,7,100,40]
# [7,10,40,100]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```

```python
# Decending order
l=[10,7,100,40]
# [100,40,10,7]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]<l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```

```python
# sorted() :- it is a function that always return a new list.
# sorted(iterableobj,[reverse=False])
# return ascending order.
l=[10,7,100,40,54,67,198,23]
l1=sorted(l)
print(l1)
```

```python
l=[10,7,100,40,54,67,198,23]
l1=sorted(l,reverse=True)
print(l1)
```

```python
l=(10,7,100,40,54,67,198,23)   # tuple
l1=sorted(l)
print(l1)        # always return a list
```

```python
# sort() :- method
# listobj.sort([reverse=False])
# by default - ascending order
l=[10,7,100,40,54,67,198,23]
l.sort()
print(l)
```

```python
l=[10,7,100,40,54,67,198,23]
l.sort(reverse=True)
print(l)
```
