```python
# Map() :- Transform every element of iterable

# Filter() :- It is used to filter some elements from iterables based on some condition
# syntax :- filter(function,iterables)

a=[1,2,3,4,5,6,7,8]
even=filter(lambda x:x%2==0,a)
print(list(even))

a=[1,2,3,4,5,6,7,8]
odd=filter(lambda x:x%2!=0,a)
print(list(odd))

a=[1,2,3,4,5,6,7,8]
num=filter(lambda x:x>3,a)
print(list(num))

a=[1,2,3,4,5,6,7,8]
num=filter(lambda x:x%3==0,a)
print(list(num))
```

```python
names=["riya","gagan","ruchi","bhavya","aman"]
l=filter(lambda x:len(x)==5,names)
print(list(l))
```

```python
names = ["Amit", "Rahul", "Ankit", "Priya", "Ajay", "Neha"]
result=filter(lambda name:name.startswith("A"),names)
print(list(result))

result=filter(lambda name:name.endswith("t"),names)
print(list(result))
```
