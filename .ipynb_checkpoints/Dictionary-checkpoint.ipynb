```python
# Dictionary :- Data store in key:value pair
# It is mutable,ordered,unindexed
# key should be unique(integer,string,tuple)
# value should be duplicate

d=dict()                      # 1st method
e={"name":"Gagan","age":32}   # 2nd method     e={key:value}
print(type(d))
print(type(e))
```

```python
# access value of dictionary
e={"name":"Gagan","age":32}
print(e["name"])
print(e["age"])

# Adding new key value pair in dictionary
e["course"]="DA"       # new key:value pair added
print(e)

# Update value of key
e["age"]=30         # update value of key
print(e)
```

```python
# value should be duplicate in dictionary
marks={"maths":90,"science":90,"english":90}
print(marks)

student={"name":"Bhavya","name":"Ruchi"}
print(student)           # if we create same key than last key overwrite

a={10:"A",20:"B",30:"C"}   # here key is integer
print(a)
```

```python
# Nested Dictionary :- dictionary inside dictionary 
# JSON format (Javascript Object Notation)

employees={
    101:{
        "name":"Bhavya",
        "salary":50000,
        "designation":"DA"
    },
    102:{
        "name":"Ruchi",
        "salary":50000,
        "designation":"DA"
    },
    103:{
        "name":"Disha",
        "salary":50000,
        "designation":"DA"
    }
}
print(employees)
```

```python
# Functions in Dictionary
# len(),max(),min(),sorted(),sum()
# element = key:value
e={"name":"Gagan","age":32,"course":"DA"}
print(len(e))   # it gives total element in dictionay
print(max(e))   # if there are not integer than compare ASCII value
print(min(e))   # if there are not integer than compare ASCII value
print(sorted(e))
print(sum(e))   # sum function only works when there are integer otherwise error occurs
```

```python
e={10:"Gagan",30:32,20:"DA"}
print("Length",len(e))   
print("Maximum",max(e))   
print("Minimum",min(e))   
print("Sum",sum(e))
print(sorted(e))                # it gives arranged list of keys
print(sorted(e,reverse=True))   # reverse list of keys
```

```python
# Methods in dictionary
# keys() :- It gives keys of dictionary
# values() :- It gives values of dictionary
# items() :- It gives both keys and values in dictionary

student={"name":"Gagan","age":32,"course":"DA"}
print(student.keys())
print(student.values())
print(student.items())
```

```python
# Looping in dictionary
student={"name":"Gagan","age":32,"course":"DA"}
for i in student.keys():
    print(i,end=" ")
print()

for i in student.values():       # to get values of dictionary using loop
    print(i,end=" ")
print()

for i in student.items():       # to get both key value of dictionary using loop,it returns output as a tuple
    print(i)              
```

```python
# get() :- It return value of key if available in dictionary 
         # otherwise return none
         # if we provide default value then it returns that default value

d={"name":"Gagan","age":32}

print(d["name"])      # 1st method to get value
print(d.get("name"))  # 2nd method to get value

print(d.get("course"))
print(d["course"])      # errror in this line because key is not available

    
```

```python
# Adding new key:value pair in dictionary 
# update() :- it is used to add single/multiple key:value pair in dictionary
d={"name":"Gagan","age":32}
print(d)
d["course"]="DA"
print(d)
a={"city":"bhopal","Qualification":"BE"}
d.update(a)
print(d)
d.update({"city":"Indore"})
print(d)
```

```python
# remove elements from dictionary
# pop(key) :- It is used to remove any particular key and value from dictionary 
            # It returns value
# popitem() :- It removes last element(key:value:)
             # It returns key and value in tuple

d={"name":"Gagan","age":32,"city":"bhopal","course":"DA"}
a=d.pop("age")
print(d)
print(a)
b=d.popitem()
print(d)
print(b)
```

```python
# del() :- It removes particular key or remove entire dictionary

d={"name":"Gagan","age":32,"city":"bhopal","course":"DA"}
del d["city"]
print(d)
del d
print(d)
```

```python
# copy() :- To create copy of dictionary 
# clear() :- To remove all element of dictionary 

d={"name":"Gagan","age":32,"city":"bhopal","course":"DA"}
a=d.copy()
print(a)
d.clear() 
print(d)
print(type(d))
```

```python
# setdefault(key,default_value)

d={"name":"Gagan","age":32,"city":"bhopal","course":"DA"}
a=d.setdefault("name","Bhavya")
print(d)
b=d.setdefault("fees",30000)     # setdefault method update dictionary if key is not in dictionary
print(d)
```

```python
# fromkeys(iterable,default_value)

subject=["Maths","Science","English"]
d=dict.fromkeys(subject,100)
print(d)
```

```python
# dictionary comprehension
result={i:i**2 for i in range(1,11)}    # key:value
print(result)
```

```python
# Question 1 create a dictionary which has number as a key and value 
# is square of that number from 1 to 10

d=dict()
for i in range(1,11):
    d[i]=i**2
print(d)
```
