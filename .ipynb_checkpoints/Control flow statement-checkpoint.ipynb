```python
# CONTROL FLOW STATEMENT
## Conditional Statement
## if block
## if block only works for true condition 

age=int(input("Enter Age : "))

if age>=18:
    print("Eligible for vote")
```

```python
## if-else block 
## if block work on True condition and else block works on False condition
age=int(input("Enter Age : "))

if age>=18:
    print("Eligible for vote")
else:
    print("Not Eligible for vote")
```

```python
## if-elif-else block
## If there are multiple True condition then we use this block

marks=int(input("Enter Marks : "))
if marks>=85:
    print("A Grade")
elif marks>=70 and marks<85:
    print("B Grade")
elif marks>=50 and marks<70:
    print("C Grade")
elif marks>=35 and marks<50:
    print("D Grade")
elif marks<0:
    print("Please Enter Valid Marks")
else:
    print("Fail")
```

```python
## nested if
## if block inside if block

age=int(input("Enter Age : "))
citizenship=input("Enter Citizenship : ")
if age>=18:
    if citizenship=="indian":
        print("You are eligible for vote")
    else:
        print("Indian Citizenship Required for vote")
else:
    print("Age should be greater than 18")
```

```python
age=int(input("Enter Age : "))
citizenship=input("Enter Citizenship : ")
if age>=18 and citizenship=="indian":
    print("You are eligible for vote")
else:
    print("You are not eligible for vote")
```

```python
## shorthand if
age=int(input("Enter Age : "))
if age>=18:print("Eligible")
```

```python
## shorthand if-else
## ternary operator
age=int(input("Enter Age : "))
result="Eligible" if age>=18 else "Not Eligible"
print(result)
```
