```python
# Function :- It is used to perform any specific task.
# why us fuctions :-
  # It increase reusability.
  # It increase Modularity.
  # It increase Maintainability.
  # Easy to debug


# Types of Functions :-
# 1. Predefined Functions
# 2. User Defined Functions
    # A. Normal Functions :-1. No argument no return type # 2. No argument with return type
# 3. Argument without return type # 4. Argument with return type

    # B. Recursive Functions
    # c. Lambda Functions

# Structure of Function 
# def function_name(parameter/arguments):      # parameter-optional
#  (space or indentation)   ----------- {body of function or logic}
# return values (optional)
```

```python
# Normal Funtions :- types:-
# 1. No argument no return type
# It this type we not pass arguments with function and also not return anything
# by default function return None

def greet():          # function creation
    print("Hello")
    
greet()               # function calling
```

```python
# 2. No argument with return type

def get_pi():
    return 3.14

result=get_pi()
print(result)

print("------")

def xyz():
    a=10
    return a

result=xyz()
print(result)
```

```python
# 3. Argument without return type function

def addition(a,b,c):     # we pass arguments
    result=a+b+c
    print(result)

addition(10,20,30)
addition(20,50,80)

x=int(input("enter number"))
y=int(input("enter number"))
z=int(input("enter number"))
addition(x,y,z)
```

```python
# 4. Argument with return type

def addition(a,b,c):     # we pass arguments
    return a+b+c

x=int(input("enter number"))
y=int(input("enter number"))
z=int(input("enter number"))
result=addition(x,y,z)
print(result)
```

```python
# Question create function to multiply 3 numbers

def multiply():
    a=int(input("enter number"))
    b=int(input("enter number"))
    c=int(input("enter number"))
    return a*b*c

result=multiply()
print(result)
```

```python
# Question :- Write a function to compare two numbers and return maximum of them
def max_of_two(a,b):
    if a>b:
        return a
    else:
        return b

result=max_of_two(10,20)
print(result)
```

```python
# Question :- Write a function to compare three numbers and return maximum of them
def max_of_three(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c

result=max_of_three(10,20,50)
print(result)
```

```python
# Question :- Write a function to compare three numbers and return maximum of them

def max_of_two(a,b):
    if a>b:
        return a
    else:
        return b

result=max_of_two(10,max_of_two(20,30))
print(result)
```
