```python
# List :- List is collection of comma separated hetergoneous data.
# It is indexed,mutable,ordered
# It is created using []

a=[]         # empty list method 1
b=list()     # empty list method 2
print(type(a))
print(type(b))

c=[10,10.5,"gagan",["a","b","c"]]
print(c)
print(c[0])
print(c[1])
print(c[2])
print(c[3])
print(c[3][0])
print(c[3][1])
print(c[3][2])
```

```python
# mutable

c=[10,10.5,"gagan",["a","b","c"]]
c[1]="Decode"
print(c)
```

```python
# Direct looping

c=[10,10.5,"gagan",["a","b","c"]]
for i in c:
    print(i)

print("----------------")

# Indirect looping

for i in range(len(c)):
    print(c[i])
```

```python
# Index position define

c=[10,10.5,["Bhavya","Aliya",["Bhopal","Delhi"]],["a","b","c"]]
print(c[2])
print(c[2][2])
print(c[2][2][1])
print(c[2][2][1][2])
```

```python
# Index position define

c=[10,10.5,["Bhavya","Aliya"],["a",["Bhopal","Delhi"],["a","b","c"]]]
print(c[3])
print(c[3][1])
print(c[3][1][0])
print(c[3][1][0][3])
```

```python
# List function and methods
# len,max,min,sort,sum

# len()--

a=[10,20,30,40,50,60]
count=0
for i in a:
    count+=1
print(len(a))
print(count)
```

```python
# max method 1

a=[10,20,30,40,50,60]
largest=0
for i in a:
    if i>largest:
        largest=i
print("largest element is : ",largest)

# method 2 -- max()

a=[10,20,30,40,50,60]
print(max(a))
```

```python
# lowest

a=[10,20,30,40,50,60]
lowest=a[0]
for i in a:
    if i<lowest:
        lowest=i
print("lowest element is : ",lowest)
```

```python
# sum

a=[10,20,30,40,50,60]
s=0
for i in a:
    s=s+i
print(s)

# method 2
a=[10,20,30,40,50,60]
print(sum(a))
```

```python
# sorted -- to arrange list 

a=[10,50,80,30,40,55,50,67,60]
print(sorted(a))
print(sorted(a,reverse=True))
```

```python
# List methods
# syntax: my_list.methodname()

# Append() :- To add element in list
# use append when we want to add single element
# add only one element 

l=[10,20,30]
l.append(40)
print(l)
```

```python
# Extend :- It is used to add multiple element in list

l=[10,20,30]
l.extend((50,60))
print(l)

print("---------")

a1=[10,20,30]
a2=[50,60]
print(a1+a2)
a1.extend(a2)
print(a1)
```

```python
# Inseted() :- when we use to add element at any specific index position
# syntax :

b=[10,20,30]
print("original :",b)
b.insert(1,22)
print("modify :",b)
```

```python
# Pop() :- To remove element from list 
# syntax :- pop(index_position)
# If there is not passed any index position then it will remove last element
# pop() remove and return element

l=[10,20,30,40,50,60]
l.pop(2)
print(l)

# method 2

l=[10,20,30,40,50,60]
r=l.pop(2)
print(l)
```

```python
# Remove :- It is used to remove particular element from list
# syntax : my_list.remove(element)
# It will remove first occurance of element

d=[10,40,50,40,60,80]
d.remove(40)
print(d)
```

```python
# Clear() :- To delete all element of list

d=[10,40,50,40,60,80]
d.clear()
print(d)
```

```python
# Index() :- To find index position of any particular element

d=[10,40,50,40,60,80]
print(d.index(40))
print(d.index(40,2))
```

```python
# Count() :- It is used to count occurance of any particular element

d=[10,40,30,40,50,60,75,60]
print(d.count(40))
print(d.count(60))
print(d.count(75))
```

```python
# Sort():- It is used to arrange list

d=[10,40,30,40,50,60,75,60]
(d.sort)
print(d)
d.sort(reverse=True)
print(d)
```

```python
# Reverse() :- It is used to reverse list

d=[10,40,30,40,50,60,75,60]
r=d.reverse()
print(d)
```

```python
# Copy() :- To make copy list

d=[10,40,30,40,50,60,75,60]
e=d.copy()
print(d)
```

```python
# enumerate() - If we want index position and value both

a=["Gagan","Ruchi","Alia","Bhavya"]
for i in enumerate(a):
    print(i)
```

```python
# zip() - It is used to combine two or more lists
# syntax - zip(list1,list2,......,listn)

a=["Gagan","Ruchi","Alia","Bhavya"]
b=["Full stack","Data Analytics","Data Science","Machine Learning"]
for i in zip(a,b):
    print(i)
```

```python
# List Comprehension
## It is short and effective way to create list from existing list in single line.

## syntax :  [expression for loop if condition]

# square of number using list comprehension

a=[1,2,3,4,5,6]
square=[i**2 for i in a]
print(square)
```

```python
b=[23,12,45,64,72,43]
even=[i for i in b if i%2==0]
print(even)
```

```python
b=[23,12,45,64,72,43]
result=["Even" if i%2==0 else "Odd" for i in b]
print(result)
```

```python
b=[23,12,45,64,72,43]
odd=[i for i in b if i%2!=0]
print(odd)
```

```python
## convert to upper case and append in new list

names=["Gagan","Ruchi","Alia","Bhavya","Laxmii"]
n=[name.upper() for name in names]
print(n)
```

```python
## convert to upper case and append in new list

names=["Gagan","Ruchi","Alia","Bhavya","Laxmii"]
new_names=[]
for name in names:
    new_names.append(name.upper())
print(new_names)
```
