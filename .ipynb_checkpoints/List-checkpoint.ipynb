```python
# List :- List is collection of comma separated hetergoneous data.
# It is indexed,mutable,ordered
# It is created using []

a=[]         # empty list method 1
b=list()     # empty list method 2
print(type(a))
print(type(b))

c=[10,10.5,"gagan",["a","b","c"]]
print(c)
print(c[0])
print(c[1])
print(c[2])
print(c[3])
print(c[3][0])
print(c[3][1])
print(c[3][2])
```

```python
# mutable

c=[10,10.5,"gagan",["a","b","c"]]
c[1]="Decode"
print(c)
```

```python
# Direct looping

c=[10,10.5,"gagan",["a","b","c"]]
for i in c:
    print(i)

print("----------------")

# Indirect looping

for i in range(len(c)):
    print(c[i])
```

```python
# Index position define

c=[10,10.5,["Bhavya","Aliya",["Bhopal","Delhi"]],["a","b","c"]]
print(c[2])
print(c[2][2])
print(c[2][2][1])
print(c[2][2][1][2])
```

```python
# Index position define

c=[10,10.5,["Bhavya","Aliya"],["a",["Bhopal","Delhi"],["a","b","c"]]]
print(c[3])
print(c[3][1])
print(c[3][1][0])
print(c[3][1][0][3])
```
