# Datatypes

```python
# Datatype in python defines the type of value stored in a variable.
# on the basis of Mutability:-
# Mutable                   # Immutable
#  list                        int
#  dictionary                  float
#   set                        string
#                              tuple
#                              frozen set
```

```python
# type() :- type function is a predefine function that return the datatype or a class of particular object.
```

```python
a=10
print(a)
print(type(a))
```

```python
a=10.5
print(a)
print(type(a))
```

```python
a=10
b=2
print(a,b)
print(type(a),type(b))
c=a/b
print(c)
print(type(c))
```

```python
# input() :- it is a pre-defined function use to take input from user at a run time.
# always take str as an input
a=input("enter any number: ")
b=input("enter any number: ")
c=a+b
print(c)
print(type(c))
```

```python
# typecasting() :- it is process of converting a variable(object) from one datatype to another datatype.
a='10'
print(a)
print(type(a))
b=int(a)
print(b)
print(type(b))
```

```python
a=int(input("enter any number: "))
b=int(input("enter any number: "))
c=a+b
print(c)
```

```python
a=float(input("enter any number: "))
b=int(input("enter any number: "))
c=a+b
print(c)
```

```python
a=float(input("enter any number: "))
b=int(input("enter any number: "))
c=a+b
print(c)       # throw error because can not convert float into integer value
```

```python
# string
s=""      # empty string
print(s)
print(type(s))
```

```python
s="hello"
print(s)
print(type(s))
print(s[0])     # indexed
s[0]="z"        # immutable
print(s)
```

```python
# list - collection of elements.
# denoted by [] with csv.
# ordered,indexed,mutable and heterogenous nature
l=[]    # empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,10,"hello",10.5]
print(l)         # ordered,heterogenous nature
print(type(l))
print(l[-1])     # indexed
l[0]=100         # mutable - it does support item assignment
print(l)
```

```python
# tuple() - collection of elements
# denoted by () with csv.
# ordered,indexed,immutable and heterogenous nature

t=()      # empty tuple
print(t)
print(type(t))
```

```python
# single valued tuple

t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,10,"hello",10.5)
print(t)        # ordered and heterogenous nature
print(type(t))
print(t[0])     # indexed
t[0]=100        # immutable
print(t)
```

```python
# dictionary - collection of elements
# denoted by {k:v},keys must be unique but value can be duplicate.
# ordered,unindexed,mutable.
d={}      # EMPTY DEICTIONARY
print(d)
print(type(d))
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
print(d["name"])      # unindexed
d["name"]="pqr"       # mutable
print(d)
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
d["Name"]="pqr"
print(d)
```

```python
d={"name":["abc","pqr","Def"],"subject":["DA","DA","DE"],"fees":30000}
print(d)
print(type(d))
print(d["name"][-1])
d["name"][-1]="xyz"
print(d)
```

```python
# set - collection of unique elements.
# denoted by {} with csv.
# unordered,unindexed,mutable and heterogenous nature
s=set()       # set() - set constructor
print(s)
print(type(s))
```

```python
s={10,20,10,20,"hello",10.5,10,20,100}
print(s)
print(type(s))
s.add(1000)    # mutable
print(s)
```

```python
# Frozenset - collection of unique elements.
# denoted by a constructor.
# unordered,unindexed,immmutable and heterogenous in nature
f=frozenset()
print(type(f))
```

```python
f=frozenset({10,20,10,20,"hello",10.5})
print(f)
print(type(f))
```

```python
l=[10,20,10,20,"hello",10.5]
t=(10,20,10,20,"hello",10.5)
s={10,20,10,20,"hello",10.5}
s1="welcome to bhopal"
print(frozenset(l))
print(frozenset(t))
print(frozenset(s))
print(frozenset(s1))
```
