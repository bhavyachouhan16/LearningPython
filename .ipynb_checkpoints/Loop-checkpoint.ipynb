```python
# While loop
# Syntax:- while condition:
    # statements
    # loop control statement
i=1
while i<=5:
    print(i)
    i+=1
```

```python
# LOOP CONTROLLING METHOD
# 1. Break:- If we want to stop loop at particular condition

for i in range(1,11):
    if i==5:
        break
    print(i)
```

```python
# 2. Continue:- It skip current iterration

for i in range(1,11):
    if i==5:
        continue
    print(i)
```

```python
# 3. Pass:- It is just a placeholder, It do nothing but
          # It is important because if we want to add some logic in future
          # then we write pass

for i in range(1,11):
    if i==5:
        pass
    print(i,end=" ")
```

```python
# Patterns

for i in range(1,6):      # Rows
    for j in range(1,6):  #Columns
        print("*",end=" ")
    print()
```

```python
# reactangle

for i in range(1,6):
    for j in range(1,6):
        if j<=i:
            print("*",end=" ")
    print()
```

```python
# Reverse

for i in range(1,6):
    for j in range(1,6):
        if j<=6-i:
            print("*",end=" ")
    print()
```

```python
# Question

for i in range(1,6):
    for j in range(1,6):
        if j>=6-i:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
