```python
# wap to check whether a given number is an even number or an odd number?
# (take number from user at runtime)
num=int(input("enter any number: "))
if num%2==0:
    print(num,"even number")
else:
    print(num,"odd number")
```

```python
# wap to check whether a user eligible to vote or not?
# (take users age from user at runtime)
age=int(input("enter age: "))
if age>=18:
    print(age,"eligible to vote")
else:
    print(age,"not eligible to vote")
```

```python
# wap to check whether a given number is divisible by 3 or not?
# (take number from user at runtime)
num=int(input("enter any number"))
if num%3==0:
    print(num,"is divisible by 3")
else:
    print(num,"is not divisible by 3")
```

```python
# wap to check whether a last digit of any number is divisible by 3 or not?
num=int(input("enter any number: "))
last_digit=num%10
if last_digit%3==0:
    print("divisible")
else:
    print("not divisible")
```

```python
# wap to check whether a given number is positive,negative or a zero?
num=int(input("enter number: "))
if num>0:
    print(num,"positive")
elif num<0:
    print(num,"negative")
else:
    print(num,"zero")
```

```python
# wap to find largest among three number?(take input from user at runtime)
num1=int(input("enter number1: "))
num2=int(input("enter number2: "))
num3=int(input("enter number3: "))
if num1 > num2 and num1 > num3:
    largest=num1
elif num2 > num1 and num2 >num3:
    largest=num2
else:
    largest=num3
print("largest among three number: ",largest)
```

```python
# wap to check whether a given number is positive ,negative or zero?
# if number is positive then also check is it even or odd?
num=int(input("enter any number: "))
if num>0:
    if num%2==0:
        print(num,"positive even number")
    else:
        print(num,"positive odd number")
elif num<0:
    print(num,"negative")
else:
    print(num,"zero")
```

```python
sub1=int(input("enter marks of sub1: "))
sub2=int(input("enter marks of sub2: "))
sub3=int(input("enter marks of sub3: "))
sub4=int(input("enter marks of sub4: "))
sub5=int(input("enter marks of sub5: "))
obt_marks=sub1+sub2+sub3+sub4+sub5
print("obtained marks: ",obt_marks)
per=(obt_marks/500)*100
print("percentage: ",per,"%")
if per>=65:
    print("First Division")
elif per>=50:
    print("Second Division")
elif per>=33:
    print("Third Division")
else:
    print("Fail")
```
