```python
# Question 1: Student Class
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

s1=Student("Bhavya",25)
print(s1.name,s1.age)
```

```python
# Question 2: Rectangle Class
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

r1=Rectangle(5,10)
print(r1.length*r1.width)
```

```python
# Question 3: Employee Class
class Employee:
    def __init__(self,emp_name,salary):
        self.emp_name=emp_name
        self.salary=salary

e1=Employee("Tilak",50000)
e2=Employee("Raj",30000)
print(e1.emp_name,e1.salary,e2.emp_name,e2.salary)
```

```python
# Question 4: Car Class
class Car:
    def __init__(self,company,model,price):
        self.company=company
        self.model=model
        self.price=price
    def showDetails(self):
        print("Car Company",self.company)
        print("Car Model",self.model)
        print("Car Price",self.price)
        
c1=Car("thar",13,2000000)
c1.showDetails()

c2=Car("sierra",12,1149000)
c2.showDetails()

c3=Car("defender",14,1070000)
c3.showDetails()
```

```python
# Question 5: Bank Account
class Bank:
    def __init__(self,balance):
        self.__balance=balance
    
    def displaybalance(self):
        print("balance is",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount

a1=Bank(50000)
a1.displaybalance()
a1.setbalance(2000)
a1.displaybalance()
```

```python
# Question 6: ATM Machine
class ATM:
    def __init__(self,pin):
        self.__pin=pin

    def showpin(self):
        print("Old Pin is",self.__pin)

    def changepin(self,newpin):
        self.__pin=newpin

    def newpin(self):
        print("New Pin is",self.__pin)

p1=ATM(6543)
p1.showpin()
p1.changepin(9876)
p1.newpin()
    
```

```python
# Question 7: Person -> Student
class Person:
    def __init__(self,name):
        self.name=name
        
    def displayPerson(self):
        print("Name is",self.name)

class Student(Person):
    def __init__(self,rollnumber):
        self.rollnumber=rollnumber
        
    def displayStudent(self):
        print("RollNumber is",self.rollnumber)
        

s=Student()
s.displayStudent()
```
