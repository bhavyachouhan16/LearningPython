```python
# OPERATOR:-1.Arithmatic 2.Assignment 3.Comparison 4.Logical 5.Membership 6.Identity 7.Bitwise
# ASSIGNMENT OPERATOR
## It is used to assign value to variable
# =,+=,-=,*=,/=,%=,**=,//=

a=10
a+=15        # a= a+15
print(a)

b=15
b-=10        # b=b-10
print(b)

c=6
c*=10        # c=c*10
print(c)

d=20
d/=5         # d=d/5
print(d)

e=10
e%=3         # e=e%3
print(e)

f=5
f**=2        # f=f**2
print(f)

g=20
g//=5         # g=g//5
print(g)
```

```python
# ARITHMATIC OPERATOR
# +,-,*,/,%,**,//

a=10
b=3
print(a+b)      #ADDITION
print(a-b)      #SUBTRACTION
print(a*b)      #MULTIPICATION
print(a/b)      #DIVISON (gives float value)
print(a//b)     #FLOOR DIVISON (remove decimal part)
print(a%b)      #MODULUS(give remainder)
print(a**b)     #EXPONANT (power)
```

```python
# COMPARISON OPERATOR
# It compare value and give True or False
# ==,!=,>,<,>=,<=

a=10
b=5
print(a==b)    #False
print(a!=b)    #True
print(a>b)     #True
print(a<b)     #False
print(a>=b)    #True
print(a<=b)    #False
```

```python
# LOGICAL OPERATOR
# and,or,not

a=10
b=20
c=30
print(a>b and b>c)    #False and False   #False
print(a<b and b>c)    #True and False    #False
print(a<b or b>c)     #True or False     #True
print(not(a>b))       #not False         #True
print(not(a<b))       #not True          #False
```

```python
# MEMBERSHIP OPERATOR:-Check Membership in different Data-types
# in , not in
# Always gives True or False

print("a" in "gagan")   #True
print("a" not in "gagan")   #False
```

```python
# IDENTITY OPERATOR 
# is,is not
# It compares the address of variable
# comparison operator(==) compare values and identity operator compare address of value

a=[10,20]
b=[10,20]
print(a==b)   #True
print(a is b)  #False
print(id(a))
print(id(b))
b=c
print(b is c)  #True
print(b is not c)  #False
```

```python
# BITWISE OPERATOR
# bin()- it is used to find binary of any number
# bitwise and -- &
# bitwise or --|
# bitwise xor -- ^
# bitwise not -- ~
# f- string - it is used to access variable in string 
a=5
b=3
print(f"binary of {a} is ",bin(a))
print(f"binary of {b} is ",bin(b))
print(a&b)  #binary AND operator
print(a|b)  #binary OR operator
print(a^b)  #binary NOT operator
print(~a)   #note:- ~n=-(n+1) beacuse ~5=-(5+1)=-6 :- python store sign binary
```
