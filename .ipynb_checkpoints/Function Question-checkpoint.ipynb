```python
# QUESTION 3 :- Find Maximum of Three Numbers
# Write a function to return the largest among three numbers.
def max_of_three(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
a=int(input("enter number"))
b=int(input("enter number"))
c= int(input("enter number"))
result=max_of_three(a,b,c)
print(result)
```

```python
# QUESTION 6 :-  Calculate Factorial
# Write a function to calculate the factorial of a number.

def factorial(n):
    if n<0:
        return "Enter Positive Number"
    else:
        fact=1
        for i in range(1,n+1):
            fact=fact*i

    return fact

print(factorial(5))
```

```python
# REVERSE 
def factorial(n):
    if n<0:
        return "Enter Positive Number"
    else:
        fact=1
        for i in range(n,0,-1):
            fact=fact*i

    return fact

print(factorial(5))
```

```python
# QUESTION 7 :-Check Prime Number
# Write a function to check whether a number is prime

def prime(num):
    if num<1:
        return "Not Prime Number"
    else:
        is_prime=True
        for i in range(2,num):
            if num%i==0:
                is_prime=False
                break
        if is_prime:
            return "Prime Number"
        else:
            return "Not Prime Number"

print(prime(12))
```

```python
# QUESTION 8 :-  Print Prime Numbers in a Given Range
# Write a function to print list of all prime numbers between two given numbers

def prime_list(start,stop):
    l=[]
    if start>stop or start<0 or stop<0:
        return "Enter Valid Range"
    else:
        for i in range(start,stop+1):
            is_prime=True
            for j in range(2,i):
                if i%j==0:
                    is_prime=False
                    break
            if is_prime:
                l.append(i)

    return l
print(prime_list(10,100))
```

```python
# QUESTION 9 :- Generate Fibonacci Series
# Write a function to generate the list of Fibonacci series up to N terms.

def feb(num):   #num=5   #o/p= 0 1 1 2 3
    a=0
    b=1
    if num<=0:
        return "Enter Positive Integer"
    else:
        if num==1:
            return [a]
        elif num==2:
            return [a,b]
        else:
            feb_list=[a,b]
            for i in range(3,num+1):       # num=3
                c=a+b                         # [0,1,1]
                feb_list.append(c)
                a=b
                b=c
        return feb_list
print(feb(5))
```

```python
# QUESTION 10 :- Reverse a Number
# Write a function to reverse the digits of a number.

def reverse(n):
    rev=0
    while n>0:
        digit=n%10
        rev=rev*10+digit
        n=n//10
    return rev
print(reverse(123))
```

```python
# QUESTION 11 :- Sum of Digits
# Write a function to find the sum of the digits of a number
# n=12345
# sum=0

def sum_num(n):
    if n==0:
        return 0
    else:
        s=0
        while n>0:
            digit=n%10
            s=s+digit
            n=n//10
    return s
print(sum_num(12345))
```

```python
# QUESTION 12 :- Count Digits
# Write a function to count the total digits in a number

def count_digit(n):
    if n==0:
        return 1
    else:
        c=0
        for i in range(len(str(n))):
            c+=1
    return c
print(count_digit(9856))
```

```python
# QUESTION 13 :- Check Palindrome Number
# Write a function to check whether a number is a palindrome

def palidrome_num(n):
    if n<=0:
        return "Not Palidrome Number"
    else:
        temp=n
        s=0
        while temp>0:
            digit=temp%10
            s=s*10+digit
            temp=temp//10
        if s==n:
            return "Palidrome Number"
        else:
            return "Not Palidrome Number"
            
print(palidrome_num(121))
```

```python
# QUESTION 14 :- Reverse a String
# Write a function to reverse a string.

def reverse_string(s):
    rev=""
    for i in s:
        rev=i+rev
    return rev
print(reverse_string("bhavya"))
```

```python
# QUESTION 15 :-  Check Palindrome String
# Write a function to check whether a string is a palindrome.

def palidrome_string(s):
    rev=""
    for i in s:
        rev=i+rev
    if s==rev:
            return "palidrome string"
    else:
            return "not palidrome sting"
print(palidrome_string("level"))
```

```python
# QUESTION 16 :- Count Vowels in a String
# Write a function to count the vowels in a string

def count_vowels(s):
    vowel_count=0
    for i in s:
        if i in "aeiou":
            vowel_count+=1
    return vowel_count
print(count_vowels("reputation"))
```

```python
# QUESTION 17 :- Find Largest Element in a List
# Write a function to return the largest element of a list.

def largest_element(num):
    largest=num[0]
    
    for i in num:
        if i > largest:
            largest=i
    return largest
num=[10,20,40,60]
print(largest_element(num))
            
```

```python
# QUESTION 18 :- Sum of List Elements
# Write a function to calculate the sum of all list elements.

def sum_list(num):
    s=0
    for i in num:
        s=s+i
    return s
num=[10,20,30,40]
print(sum_list(num))
```

```python
# QUESTION 19 :- Calculator using Functions
 #     Create separate functions for:
  #      -   Addition
   #     -   Subtraction
    #    -   Multiplication
     #   -   Division
      #  -   Modulus

def addition(a,b,c):    
    result=a+b+c
    print(result)

addition(10,20,30)

print("-----")

def subtraction(a,b,c):     # we pass arguments
    result=a-b-c
    print(result)

subtraction(10,2,3)

print("------")

def multiplication(a,b,c):     # we pass arguments
    result=a*b*c
    print(result)

multiplication(2,4,5)

print("-----")

def division(a,b):     # we pass arguments
    result=a/b
    print(result)

division(90,3)

print("----")

def modulus(a,b):     # we pass arguments
    result=a%b
    print(result)

modulus(10,3)
```

```python
# QUESTION 20 :- Student Grade Calculator
# Write a function to calculate the grade based on marks.

def stu_grade(mark):
    if mark>=90:
        return "A"
    elif mark>=80:
        return "B"
    elif mark>=60:
        return "C"
    else:
        return "D"
print(stu_grade(77))
```

```python
# QUESTION 21 :- Check Armstrong Number
# Write a function to check whether a number is a Armstrong.

def armstrong_check(num):
    if num==0:
        return 0
    else:
        s=0
        temp=num
        power_of_digit=len(str(num))
        while temp>0:
            digit=temp%10
            s=s+digit**power_of_digit
            temp=temp//10
    if s==num:
            return "Armstrong Number"
    else:
            return "Not Armstrong Number"
print(armstrong_check(153))
```
