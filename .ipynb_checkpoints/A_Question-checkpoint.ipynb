```python
# Question 1 largest of three number
a=int(input("enter any number: "))
b=int(input("enter any number: "))
c=int(input("enter any number: "))
if a>b and a>c:
    print(a,"is greater")
elif b>a and b>c:
    print(b,"is greater")
else:
    print(c,"is greater")
```

```python
# Question 2 leap year
year=int(input("enter year: "))
if year%4==0 and year%100!=0 or year%400==0:
    print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
# Question 3 number multiple by 3 and 7 both
num=int(input("enter any number: "))
if num%3==0 and num%7==0:
    print(num,"is divisible by both 3 and 7")
else:
    print(num,"is not divisible by both 3 and 7")
```

```python
# Question 4 student grade
student_marks=int(input("enter any number: "))
if student_marks>=0 and student_marks<=100:
    if student_marks>=90:
        print(student_marks,"A+")
    elif student_marks>=80:
        print(student_marks,"A")
    elif student_marks>=70:
        print(student_marks,"B")
    elif student_marks>=60:
        print(student_marks,"C")
    elif student_marks>=40:
        print(student_marks,"D")
    else:
        print(student_marks,"Fail")
else:
    print("Invalid Marks")
```

```python
# Question 5 student scholarship eligibility
Marks=int(input("enter any number: "))
Att_per=int(input("enter any number: "))
if Marks>=85 and Att_per>=75:
    print("student is eligible for the scholarship")
else:
    print("student is not eligible for the scholarship")
```

```python
# Question 6 salary bonus calculator
salary=int(input("enter any number: "))
if salary>=50000:
    bonus=salary*10/100
    salary=salary+bonus
    print(bonus,"salary after 10% bonus: ",salary)
else:
    bonus=salary*5/100
    salary=salary+bonus
    print(bonus,"salary after 5% bonus: ",salary)
```

```python
# Question 7 Electricity bill calculator
unit=int(input("enter unit: "))
if unit>0:
    if unit<=100:
        per_unit=unit*5
    elif unit<=200:
        per_unit=100*5+(unit-100)*7
    else:
        per_unit=100*5+100*7+(unit-200)*10
else:
    print("invalid")
print(per_unit)
```

```python
# Question 9 Simple Calculator 
a=int(input("enter any number: "))
b=int(input("enter any number: "))
oper=input('''
            1. +
            2. -
            3. *
            4. /
            ''')
if oper=="1":
    print(a+b)
elif oper=="2":
    print(a-b)
elif oper=="3":
    print(a*b)
elif oper=="4":
    if b!=0:
        print(a/b)
    else:
        ("error")
else:
    ("invalid operator")
```

```python
# Question 10 Check Whether a Character Is an Alphabet
ch=input("enter any character: ")
if len(ch)==1:
    if (ch>='A' and ch<='Z') or (ch>='a' and ch<='z'):
        print(ch,"is an alphabet")
    else:
        print(ch,"is not an alphabet")
else:
    print("enter only one character")
```

```python
# Question 11 Vowel or Consonant
ch=input("enter any character: ")
if len(ch)==1:
    if (ch>='A' and ch<='Z') or (ch>='a' and ch<='z'):
        if ch in "aeiouAEIOU":
            print(ch,"is vowel")
        else:
            print(ch,"is consonant")
    else:
        print("invalid input")
else:
    print("enter only one character")
```

```python
# Question 12 ATM Withdrawal System
current_balance=float(input("enter current balance: "))
withdrawal_amount=float(input("enter withdrawal amount: "))
if withdrawal_amount>0 and current_balance>=withdrawal_amount:
        current_balance=current_balance-withdrawal_amount
        print("transaction successful")
        print("updated balance: ",current_balance)
else:
    print("withdrawl amount greater than 0 and less than or equal to current balance")
```
