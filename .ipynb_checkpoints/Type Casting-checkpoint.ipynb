```python
# Type Casting
# variable's datatype change
# Implicit (run type)
# Explicit (manually)

# implicit
a=10
b=12.5
c=a+b
print(c)
```

```python
a=1
b=True
c=a+b
print(c)
```

```python
# Type casting
## any other form to integer
## integer to any form

a="123"
b=10.5   #float value
print(int(b))   #float to integer

c=10
print(float(c))       # integer to float
print(type(str(c)))   # integer to string
print(list(c))       # error here because integer is immutable
```

```python
## tuple to list..... pass tuple in list() as argument

a=[10,"Gagan",10.5]
t=tuple(a)          # list to tuple
print(t)
print(type(t))

b=(10,"gagan",10.5)
l=list(b)           # tuple to list
print(l)
print(type(l))
```

```python
# list to set

a=[10,10,20,25.5,20,"Gagan","gagan"]
# change it to set
# print that set and their dataype
s=set(a)
print(s)
print(type(s))
print(id(s))
print(len(a))
print(bin(2))
```
