```python
# Polymorphism :- Poly means Many and Morphism means Forms
# Types of Polymorphism :-
# 1. Over-Riding :- a. Method over-riding
# 2. Overloading :- a. Method overloading   b. Operator overloading

# Method Over-riding :- In Python,If same method present in parent and child class
                      # and we create object of child class then that method in child class
                      # override parent class's method.

class Parent:
    def display(self):
        print("Its Parent Class")

class Child(Parent):
    def display(self):
        print("Its Child Class")

c=Child()
c.display()        # chlid class's display method override parent's display method.
```

```python
# Overloading
# 1. Method Overloading :- Not supported in Python

def add(a,b):
    print(a+b)
add(1,2)

def add(a,b,c):
    print(a+b+c)
add(1,2,3)
```

```python
# Method Overloading:- n number of arguments(args)

def add(*args):
    print(sum(args))      # args = tuple()

add(1,2)
add(1,2,3)
add(1,2,3,4,5,6,7,8,9)
```

```python
# 2. Operator Overloading :- 

class Employee:
    def __init__(self,name,salary):      # Dunder Method or Special Method for operator overloading
        self.name=name
        self.__salary=salary

    def display(self):
        print("Employee name is ",self.name,"And Salary is",self.__salary)

    def __add__(self,other):      # Dunder Method     # Double Underscore Method
        return self.__salary+other.__salary

e1=Employee("Yash",50000)
e1.display()

e2=Employee("Saurav",80000)
e2.display()

print(e1+e2)       # e1.__add__(e2)   # e1.salary+e2.salary
```
