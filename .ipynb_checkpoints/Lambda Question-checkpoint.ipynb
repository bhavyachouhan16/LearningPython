```python
#QUESTION 1: ADD TWO NUMBERS
#Create a lambda function that takes two numbers and returns their sum.

addition=lambda a,b:a+b
print(addition(10,5))
```

```python
#QUESTION 2: SUBTRACT TWO NUMBERS
#Create a lambda function that takes two numbers and returns the result
#of the first number minus the second number.

subtraction=lambda a,b:a-b
print(subtraction(90,30))
```

```python
#QUESTION 3: MULTIPLY TWO NUMBERS
#Create a lambda function that takes two numbers and returns their multiplication

multiplication=lambda a,b:a*b
print(multiplication(5,6))
```

```python
#QUESTION 4: DIVIDE TWO NUMBERS
#Create a lambda function that takes two numbers and returns their division.

division=lambda a,b:a/b
print(division(90,3))
```

```python
#QUESTION 5: FIND THE SQUARE OF A NUMBER
#Create a lambda function that accepts one number and returns its square.

square=lambda a:a*a
print(square(6))
```

```python
#QUESTION 6: FIND THE CUBE OF A NUMBER
#Create a lambda function that accepts one number and returns its cube.

cube=lambda a:a*a*a
print(cube(3))
```

```python
#QUESTION 7: FIND THE REMAINDER
#Create a lambda function that accepts two numbers and returns the
#remainder when the first number is divided by the second number

remainder=lambda a,b:a%b
print(remainder(10,3))
```

```python
#QUESTION 8: FIND THE QUOTIENT
#Create a lambda function that accepts two integers and returns the
#integer quotient of their division.

quotient=lambda a,b:a//b
print(quotient(10,4))
```

```python
#QUESTION 9: ADD 10 TO A NUMBER
#Create a lambda function that accepts a number and adds 10 to it.

num=lambda a=int(input("enter any number")):a+10
print(num(a))
```

```python
#QUESTION 10: CONVERT CELSIUS TO FAHRENHEIT
#Create a lambda function to convert temperature from Celsius to Fahrenheit.

fer=lambda cel:cel*9/5+32
print(fer(32))
```

```python
#QUESTION 11: CALCULATE SIMPLE INTEREST
#Create a lambda function to calculate Simple Interest

si=lambda p,r,t:p*r*t/100
print(si(5000,10,2))
```

```python
#QUESTION 12: CALCULATE AREA OF A RECTANGLE
#Create a lambda function that accepts length and width and returns
#the area of a rectangle.

area_rect=lambda a,b:a*b
a=int(input("enter number"))
b=int(input("enter number"))
print(area_rect(a,b))
```

```python
#QUESTION 13: CALCULATE PERIMETER OF A RECTANGLE
#Create a lambda function that accepts length and width and returns
#the perimeter of a rectangle.

peri_rect=lambda a,b:2*(a+b)
a=int(input("enter number"))
b=int(input("enter number"))
print(peri_rect(a,b))
```

```python
#QUESTION 14: JOIN FIRST NAME AND LAST NAME
#Create a lambda function that accepts a first name and last name and
#returns the complete name.
join_name=lambda name,surname:name+" "+surname
f=input("Enter First Name")
l=input("Enter Last Name")
print(join_name(f,l))
```

```python
#QUESTION 15: CALCULATE TOTAL PRICE
#Create a lambda function that accepts price and quantity and returns
#the total price.

total_price=lambda q,p:q*p
q=int(input("enter quantity"))
p=int(input("enter price"))
print(total_price(q,p))
```
