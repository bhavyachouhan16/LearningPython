```python
# SET
## created using {},it is comma separated values
## heterogenious(multiple data)
## mutuable , unordered , unindexed
## duplicate values not allowed

s={10,10.5,"Gagan"}
print(type(s))
print(s)

a=[10,20,30,20,20,"gagan","gagan"]     # list 
print(set(a))

print("------")

# remove duplicate values from list using typecasting(by set())

s=set(a)
l=list(s)
print(l)
```

```python
# looping through set
a={10,20,30,20,20,"gagan","gagan"}
for i in a:
    print(i)
```

```python
# Functions of set
# len(),max(),min(),sum(),sorted()
# sorted() :- return output as a list

a={10,10,20,30,30,30,15,5}
print(len(a))
print(max(a))
print(min(a))
print(sum(a))
print(sorted(a))
print(sorted(a,reverse=True))
```

```python
# Methods of Set
# Adding element in set
## Add() :- It is used to add single element
## Update() :- It is usedn to add multiple element

a={1,2,3,4,5}
a.add(10)             # add single element
print(a)
a.update({100,200})   # add multiple element
print(a)
```

```python
# Remove element from Set 
## pop() :- It removes random element from set
## remove() :- It removes any particular element from set,If element not found 
            #  then throw error
## discard() :- It removes any particular element from set,If element not found 
              # then not throw any error

a={1,2,3,4,5,6,7}
a.pop()
print(a)
a.remove(6)
print(a)
a.discard(4)
print(a)
a.discard(10)        # not throw any error if element not found
print(a)
a.remove(10)         # throw error if element not found
```

```python
# copy() :- To create copy of set
# clear() :- To remove all elements of set
# del() :- To delete set from memory

a={1,2,3,4,5,6,7}
b=a.copy()     #create copy of set
print(b)
print(id(a))
print(id(b))
a.clear()     # clear elements from set
print(a)
del a          # delete set
print(a)
```

```python
## Set Operation
# union() :- join two or more set 
  # syntax :- set1.union(set2) //// set1 | set2
# intersection()
# difference()
# symmetic_difference()

s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.union(s2)
a=s1 | s2
print(s)
print(a)
```

```python
# intersection() :- Return common element
  # syntax :- set1.intersection(set2)  /// set1 & set2

s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.intersection(s2)
a=s1&s2
print(s)
print(a)
```

```python
# difference() :- Return unique element of first set
   # syntax :- set1.difference(set2) /// set1-set2
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.difference(s2)
a=s1-s2
print(s)
print(a)
```

```python
# symmetic_difference() :- Return unique element of both set
   # Syntax :- set1.symmetric_difference(set2) /// set1^set2

s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.symmetric_difference(s2)
a=s1^s2
print(s)
print(a)
```

```python
# isdisjoint :- s1.isdisjoint(s2)
s1={1,2,3,4,5}
s2={4,5,6,7,8}
print(s1.isdisjoint(s2))
```

```python
# issubset() :- s1.issubset(s2)
s1={1,2,3}
s2={1,2,3,4,5,6,7,8}
print(s1.issubset(s2))
```

```python
# issuperset() :- s2.issuperset(s1)  /// (s2 parent table s1 child table)
s1={1,2,3}
s2={1,2,3,4,5,6,7,8}
print(s2.issuperset(s1))
```
