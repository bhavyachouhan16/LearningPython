```python
# QUESTION 1: FILTER EVEN NUMBERS
numbers = [10, 15, 22, 31, 40, 51, 60]
even=filter(lambda x:x%2==0,numbers)
print(list(even))
```

```python
# QUESTION 2: FILTER ODD NUMBERS
numbers = [11, 20, 33, 44, 55, 68, 71]
odd=filter(lambda x:x%2!=0,numbers)
print(list(odd))
```

```python
# QUESTION 3: NUMBERS GREATER THAN 50
numbers = [20, 55, 32, 80, 45, 90, 12]
greater=filter(lambda x:x>50,numbers)
print(list(greater))
```

```python
# QUESTION 4: NUMBERS LESS THAN 30
numbers = [10, 45, 25, 60, 18, 35, 5]
less=filter(lambda x:x<30,numbers)
print(list(less))
```

```python
# QUESTION 5: POSITIVE NUMBERS
numbers = [-10, 20, -5, 30, 0, -15, 40]
positive=filter(lambda x:x>0,numbers)
print(list(positive))
```

```python
# QUESTION 6: NEGATIVE NUMBERS
numbers = [-10, 20, -5, 30, 0, -15, 40]
negative=filter(lambda x:x<0,numbers)
print(list(negative))
```

```python
# QUESTION 7: NUMBERS DIVISIBLE BY 5
numbers = [10, 13, 25, 32, 40, 44, 55, 61]
divi_5=filter(lambda x:x%5==0,numbers)
print(list(divi_5))
```

```python
# QUESTION 8: NUMBERS DIVISIBLE BY 3
numbers = [10, 12, 17, 21, 25, 30, 41]
divi_3=filter(lambda x:x%3==0,numbers)
print(list(divi_3))
```

```python
# QUESTION 9: NAMES STARTING WITH "A"
names = ["Amit", "Rahul", "Ankit", "Priya", "Ajay", "Neha"]
name=filter(lambda x:x.startswith("A"),names)
print(list(name))
```

```python
# QUESTION 10: NAMES WITH MORE THAN 5 CHARACTERS
names = ["Amit", "Rahul", "Alexander", "Priya", "Gagan", "Rohit"]
length_5=filter(lambda x:len(x)>5,names)
print(list(length_5))
```

```python
# QUESTION 11: NAMES ENDING WITH "A"
names = ["Riya", "Amit", "Pooja", "Rahul", "Neha", "Gagan"]
end=filter(lambda x:x.endswith("a"),names)
print(list(end))
```

```python
# QUESTION 12: WORDS WITH EXACTLY 4 CHARACTERS
words = ["book", "pen", "tree", "apple", "code", "cat"]
length=filter(lambda x:len(x)==4,words)
print(list(length))
```

```python
# QUESTION 13: FILTER NON-EMPTY STRINGS
words = ["Python", "", "Django", "", "React", "SQL"]
string=filter(lambda x:x!="",words)
print(list(string))
```

```python
# QUESTION 14: FILTER PASSING MARKS
marks = [35, 67, 22, 89, 40, 38, 76]
passing_marks=filter(lambda x:x>=40,marks)
print(list(passing_marks))
```

```python
# QUESTION 15: MARKS ABOVE 75
marks = [55, 78, 92, 65, 88, 45, 76]
marks_75=filter(lambda x:x>75,marks)
print(list(marks_75))
```

```python
# QUESTION 16: PRICES BELOW 500
prices = [250, 750, 450, 1200, 350, 900, 150]
price=filter(lambda x:x<500,prices)
print(list(price))
```

```python
# QUESTION 17: SALARIES ABOVE 50,000
salaries = [35000, 55000, 48000, 75000, 62000, 42000]
salary=filter(lambda x:x>50000,salaries)
print(list (salary))
```

```python
# QUESTION 18: FILTER ADULT AGES
ages = [12, 18, 25, 15, 32, 17, 21]
age=filter(lambda x:x>=18,ages)
print(list(age))
```

```python
# QUESTION 19: FILTER LONG WORDS
words = ["Python", "Django", "Programming", "SQL", "JavaScript", "AI"]
long_word=filter(lambda x:len(x)>6,words)
print(list(long_word))
```

```python
# QUESTION 20: NUMBERS BETWEEN 20 AND 50
numbers = [10, 22, 35, 50, 65, 28, 49, 75]
number_between=filter(lambda x:x>20 and x<50,numbers)
print(list(number_between))
```
